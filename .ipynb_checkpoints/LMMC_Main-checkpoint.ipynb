```python
from sage.rings.continued_fraction import convergents
n = k-2   ## Degree of polynomials
m = n/2
QF1.<rd1> = QuadraticField(D1)   ##Quadratic Field of D1
QF2.<rd2> =QuadraticField(D2)     ## Quadratic Field of D2
K.<prime_gen> = Qq(p**2, prec = pprec, type = 'floating-point')   ## Unramified Quadratic Extension of Qp
R.<x> = PolynomialRing(K,"x")       ## Ring of Polynomials in K
hom1 = QF1.hom([K(D1).square_root()])    ## Embedding of QF1 into K
hom2 = QF2.hom([K(D2).square_root()])     ## Embedding of QF2 into K
aut1 = QF1.automorphisms()[1]              ## Galois Conjugation for QF1
aut2 = QF2.automorphisms()[1]               ## Galois Conjugation for QF2
Pn.<T> = K['T']
from sage.rings.polynomial.polydict import ETuple
PS.<v> = TateAlgebra(K)
PU.<u> = PowerSeriesRing(K)
SavedSeries = [PS(1/(1-p*i*v)) for i in range(p)]
SavedSeriesInfty = [PS(1/(p*v-i))for i in range(1,p)]
```

```python
def unimodular_path(r):
    """
    Finds the matrices descriming a unimodular path from infty to a rational number
    :param r: a rational number
    :return: list of matrices
    """

    conv = convergents(r)
    answer = [matrix([[1,conv[0].numerator()],[0,1]])]
    for i in range(len(conv)-1):
        answer.append(matrix([[conv[i].numerator(), (-1)**(i+1)*conv[i+1].numerator()],[conv[i].denominator(), (-1)**(i+1)*conv[i+1].denominator()]]))
    return answer

Hecke = [matrix([[0,0],[0,0]]) for a in range(p+1)]
for a in range(p):
    Hecke[a] = matrix([[p,a],[0,1]])
Hecke[-1] = matrix([[1,0],[0,p]])


paths = list(range(p))
for a in range(p):
    paths[a] = unimodular_path(-a/p)
paths.append([matrix([[1,0],[0,1]])])

Sm = matrix([[0,-1],[1,0]])
Tm = matrix([[1,1],[0,1]])
Um = matrix([[0,1],[-1,1]])
```

```python
def integral_minpoly(r):
    """
    """
    return r.minpoly()/gcd((r.minpoly().coefficients()))


def mobius(x,M):
    """
    Mobious Action on a point
    :param x: A number (an element of any field)
    :param M: A 2x2 Matrix
    :return: a number in the same field as x
    """

    return((M[0][0]*x+M[0][1])/(M[1][0]*x+M[1][1]))

def ReducedForms(D):
    """
    Returns the set of reduced forms for a given discriminant. Note that for indefinite forms there isn't a unique reduced representative
    :param D: A discriminant
    :return: A list of BinaryQF
    """
    answer = []
    for R in BinaryQF_reduced_representatives(D, proper = True):
        answer += R.cycle(proper = True)
    return answer

class NegCntFraction:

    """
    A class representing continued fractions but with minus signs instead of plus. Designed for periodic continued fractions of RM points

    attribute period: The periodic part of the continued fraction
    """
    def __init__(self, preperiod, period):
        self.preperiod = preperiod
        self.period = period


def neg_continued_fraction(x):
    """
    Calculates the negative continued fraction for a given real number. Goes up to 100 digits
    :param x: a real number
    :return: NegCntFraction
    """
    answer = []
    prev = [x]
    for i in range(100):
        c = ceil(x)
        answer.append(c)
        x = 1/(c-x)

        if x in prev:
            index = prev.index(x)
            break

        prev.append(x)
    preperiod = answer[:index]
    period = answer[index:]
    return NegCntFraction(preperiod, period)


def cyclic_perm(l1,l2):
    """
    Tests if two lists are cyclic permutations of each other
    :param l1: list
    :param l2: list
    :return: Bool
    """
    l = l1 + l1
    for i in range(len(l1)):
        if l2 == l[i:i+len(l1)]:
            return True
    return False

def SymbRoots(Q):
    """
    Returns the symbolic roots of a Binary Quadratic Form in the order with the first being the + root in the quadratic formula the second being the -. Symbolic because symbolic versions of sqrt()

    :param Q: BinaryQF
    :return: An array with the two roots of the binary qf in order

    """
    a,b,c = Q[0],Q[1],Q[2]
    return [(-b+sqrt(Q.discriminant()))/(2*a),(-b-sqrt(Q.discriminant()))/(2*a)]

def RealRoots(Q):
    """
    Returns the real roots of a Binary Quadratic Form in the order with the first being the + root in the quadratic formula the second being the -.

    :param Q: BinaryQF
    :return: An array with the two roots of the binary qf in orde
    """
    return [RR(r) for r in SymbRoots(Q)]

def NegRoot(Q):
    """
    Returns the negative real root of a nearly reduced binary quadratic form
    :param Q: A nearly reduced BinaryQF
    :return: A negative real number which is a root of the QF
    """
    r,s = RealRoots(Q)
    if r>0:
        return s
    else:
        return r

def PosRoot(Q):
    r,s = RealRoots(Q)
    if r>0:
        return r
    else:
        return s

def padicRoots(Q):
    a, b, c = Q[0], Q[1], Q[2]
    return (K(a)*x**2 + K(b)*x + K(c)).roots()

def MobiusEquiv(t,z):
    """
    Determines if two elements of a quadratic field are equivalent under a mobius transformation of SL2(Z)
    :param t: NumberFieldElement_quadratic_sqrt
    :param z: NumberFieldElement_quadratic_sqrt
    :return: Bool
    """
    return cyclic_perm(neg_continued_fraction(t).period , neg_continued_fraction(z).period)

def QForm(t):
    """ Inputs an element of a quadratic field Returns a BinaryQF which has t as a root
    :param t: NumberFieldElement_quadratic_sqrt
    :return: BinaryQF
    """

    return BinaryQF(list(integral_minpoly(t))[::-1])

def automorph(t):
    """
    Computes the fundamental stabilizer associated to an RM-point.ww
    :param t: A real quadratic number given symbolically or as an element of a real quadratic field
    :return: a matrix
    """
    answer = matrix([[1,0],[0,1]])
    CF = neg_continued_fraction(t)
    for i in CF.preperiod:
        answer = Sm*Tm**(-i)*answer
    for i in CF.period:
        answer = Sm*Tm**(-i)*answer
    for i in CF.preperiod[::-1]:
        answer = Tm**i*Sm*answer
    return answer

def PosQForm(t):
    """
    Computes the binary quadratic form which t is the plus root of in the quadratic formula
    :param t: A quadratic square root
    :return: BinaryQF()
    """
    Q = QForm(t)
    if t == a.parent()(SymbRoots(Q)[0]):
        return Q
    else:
        return BinaryQF([-i for i in list(Q)])

def PolyAction(P,M):
    """
    Computes the weight -n action of a matrix on a polynomial
    :param P: a polynomial
    :param M: a 2x2 matrix
    :return: a polynomial
    """
    if (n==0): return P
    return (P(T = ((M[0][0]*T + M[0][1])/(M[1][0]*T+M[1][1])))*(M[1][0]*T+M[1][1])**n)

def PolAction(P,M):
    """
    Computes the weight -n action of a matrix on a polynomial
    :param P: a polynomial
    :param M: a 2x2 matrix
    :return: a polynomial
    """
    if n==0: return(P)
    return g(M.determinant()**(-n)*P(y = ((M[0][0]*y + M[0][1])/(M[1][0]*y+M[1][1])))*(M[1][0]*y+M[1][1])**n)

def automorph_sequence_pos(t):
    """
    """
    sequence = []
    answer = matrix([[1,0],[0,1]])
    period = continued_fraction(t).period()
    for i in range(len(period))[::-1]:
        answer = answer*Sm*Tm**((-1)**(i+1)*period[i])
        sequence.append(answer)
    return(sequence)
```

```python
def NearlyRedForm(Q):
    """
    Returns all the nearly reduced form that are equivalent to a given reduced form. (They are all translates of the original reduced form)
    :param Q: a reduced BinaryQF
    :return: a list of the reduced BinaryQFs which are equivalent to Q.
    """
    answer = []
    for PQ in Q.cycle(proper = True):
        big_root = SymbRoots(PQ)[1]
        t = int(abs(big_root))+1
        for i in range(t):
            NPQ = PQ.matrix_action_right(matrix([[1,sign(big_root)*i], [0,1]]))
            answer.append(NPQ)
            if i!= 0 and i!=(t-1):
                answer.append(NPQ.matrix_action_right(Sm))
    return answer

def NearlyRedDisc(D):
    """
     Returns all the nearly reduced form that have discriminant D. (They are all translates of the reduced forms)
    :param D: a discriminant
    :return: a list of the reduced BinaryQFs which have discriminant D. """
    answer = []
    for Q in  ReducedForms(D):
        answer += NearlyRedForm(Q)
    return answer

def NearRedFormsEquiv(t):
    """
    Finds all nearly reduced forms which are equivalent to the quadratic form of t
    :param t: a real quadratic point
    :return: a list of binary quadratic forms
    """
    answer  = []
    Q = QForm(t)
    for R in ReducedForms(Q.discriminant()):
        if Q.is_equivalent(R, proper  = True):
            answer += NearlyRedForm(R)
    return answer

def NearRedPointsEquiv(t):
    """
    Finds all real quadratic points which are an Sl2(Z) translate of t and for which its Galois conjugate is the opposite real sign
    :param t: a real quadratic point
    :return: a list of real quadratic points
    """
    answer = []
    for R in NearRedFormsEquiv(t):
        L = SymbRoots(R)
        for i in range(2):
            if MobiusEquiv(t, t.parent()(L[i])):
                answer.append(L[i])
    return answer

def NearRedPFEquiv(t):
    """
    Finds all nearly reduced forms equivalent to the form of t and returns the form and the root of the form which is equivalent to t
    :param t: a real quadratic point
    :return: a list of lists [r,R] where R is a BinaryQF and r is one of its roots
    """
    answer = []
    for R in NearRedFormsEquiv(t):
        L = SymbRoots(R)
        if MobiusEquiv(t, t.parent()(L[0])):
            answer.append([L[0],R])

    return answer

```

```python
class PolyLog:

    """
    Represents functions of the form (polynomial*log(T - point)) + rational function

    This is used to represent the Level 0 part of the cocycles i.e the parts that have residues on the standard affinoid.

    Attributes:
        point_dict: Dictionary of points as keys and polynomials as values. Represents sums of the poly * log( T- point)
        rat_function:  A rational function in T

    """
    def __init__(self, point_dict, rat_function = Pn(0)):

        """
        Initiaizes a PolyLog Object given a dictionary of points and polynomials. The rational function is automatically set to 0
        """
        self.point_dict = point_dict
        self.rat_function = rat_function

    def derivative(self):

        """
        Differentiates the PolyLog as a function
        """
        answer_dict = {}
        new_rat_function = derivative(self.rat_function)
        for point, point_functions in self.point_dict.items():

            new_rat_function += point_functions / (T-hom1(point))
            answer_dict[point] = derivative(point_functions)
        return PolyLog(answer_dict, new_rat_function)

    def __rmul__(self, scalar):
        answer_dict = {}
        for point, point_functions in self.point_dict.items():
            answer_dict[point] = scalar * point_functions
            new_rat_function = scalar * self.rat_function
        return PolyLog(answer_dict, new_rat_function)
    def __sub__(self, PL2):
        return (self + (-1) * PL2)
    def evaluate(self, s):

        """
        Evaluates the PolyLog at point
        """
        total = 0
        total += self.rat_function(s)
        for point, point_functions in self.point_dict.items():

            if n!=0:
                total += (s - hom1(point)).log(0) * point_functions(s)

            if n == 0:
                total += (s - hom1(point)).log(0) * point_functions(0)


        return total

    def __add__(self, PL2):

        """
        Adds two PolyLog functions
        """
        answer = {}
        for i in set(self.point_dict.keys()).union(set(PL2.point_dict.keys())):
            new_functions = self.point_dict.get(i,Pn(0)) + PL2.point_dict.get(i,Pn(0))

            answer[i] = new_functions
            if PL2.rat_function == 0:
                new_rat_function = (self.rat_function)
            elif self.rat_function == 0:
                new_rat_function = PL2.rat_function

            else: new_rat_function = (self.rat_function) + (PL2.rat_function)
        return PolyLog(answer, new_rat_function)

    def mobius_action(self, M):

        """
        Weight (-n) action on the polylog by an invertible 2x2 matrix.
        """
        new_dict = {}
        (a,b), (c,d) = list(M)

        if self.rat_function == 0:
            new_rat_function = Pn(0)
        else:
            new_rat_function = (c*T+d)**n * (self.rat_function)((a*T+b)/(c*T+d) )

        for point, point_functions in self.point_dict.items():

            new_point = QF1((d*point - b)/(-c*point + a))
            new_poly = PolyAction(point_functions, M)
            new_rat_function -= new_poly * hom1(c*new_point + d).log(0)
            new_dict[new_point] = new_poly


        return PolyLog(new_dict, rat_function = new_rat_function)

    def fix_poly_invariance(self):
        """
        For weight 4 specifically, the PolyLogs are often period functions up to polynomials and in certain cases this polynomial ambiguity can be fixed.

        This function fixes the polynomial ambiguity
        """
        total2 = Pn(0)
        total3 = Pn(0)

        total2 = (self.mobius_action(Sm) + self).rat_function

        total3 = (self.mobius_action(Um**2) + self.mobius_action(Um) + self).rat_function

        a2 = total2.derivative().derivative()/2
        a3 = total3.derivative().derivative()/2

        c = a2
        b = a3 - 2 * c

        self.rat_function -=  (b * T + c)/Pn(1)
```

```python
def Level0PolyLog(t):
    """
    Defines the level 0 polylogarithm associated to an RM poingt t. This is really a primitive to the rational period functions which have divisor on t
    """
    answer = PolyLog({})
    for [r, R] in NearRedPFEquiv(t):
        answer += PolyLog({QF1(r):sign(RR(r))*(R(T,1)/hom1((QF1(sqrt((R.discriminant())))) ))**m})
    return answer
```

```python
g.<y> = PolynomialRing(QQ,"y")

class Divisor():
    """
    Class to store divisors and to manipulate them. Divisors here typically are supported on RM points and take values in spaces of polynomials
    """

    def __init__(self, DivDict):
        self.DivDict = DivDict

    def DivElements(self, nz = False):

        """
        Returns the elements whose coefficients in the divisor are non-zero
        """
        if (nz == False):
            return list(self.DivDict.keys())
        elif (nz == True):
            answer = []
            for DivEl in self.DivDict:
                if self.DivDict[DivEl] != 0 :
                    answer.append(DivEl)
            return answer

    def __add__(self,Div2):

        """
        Adds two divisors
        """
        answer = {}
        for i in set(self.DivDict.keys()).union(set(Div2.DivDict.keys())):
            answer[i] = self.DivDict.get(i,0) + Div2.DivDict.get(i,0)
        newDiv = Divisor(answer)
        newDiv.clear()
        return newDiv


    def mobius_action(self, M):

        """
        Weight (-n) action on divisors
        """
        answer = {}
        for el, mul in self.DivDict.items():
            answer[mobius(el,M)] = PolAction(mul, M**(-1))
        return Divisor(answer)

    def clear(self):

        """
        Removes all the elements of the divisor whose coefficient is 0
        """
        for el in list(self.DivDict.keys()):
            if self.DivDict[el] == 0 :
                del self.DivDict[el]
    def degree(self):

        """
        Returns the degree of the divisor i.e the sum of all the multiplicities
        """
        answer = 0
        for el, mul in self.DivDict.items():
            answer += mul
        return answer

    def __rmul__(self, scalar):
        answer = {}
        for el, mul in self.DivDict.items():
            answer[el] = scalar*mul
        return Divisor(answer)
    def __sub__(self, Div2):
        return self + ((-1)*Div2)

    def __str__(self):
        return(str(self.DivDict))

    def __repr__(self):
        return(str(self.DivDict))


```

```python
def Level0Div(t):
    """
    Returns the level 0 divisor associated to an RM-point (t). This is just the set of nearly-reduced points which are equivalent to t and the multiplicities are the appropriate power of their minimal polynomial
    """

    lvl0 = Divisor({})
    for [r,R] in NearRedPFEquiv(t):
        lvl0 += Divisor({QF1(r):sign(r)*R(y,1)**m})
    return lvl0

def Level1Div(t,a):
    """
    Return the level 1 divisor associated to an RM-point whose residues are all a(mod p). More precisely, these are the nearly reduced points equivalent to any of the Hecke conjugates of t by a matrix of det p which also have reside a(mod p )
    """
    answer = Divisor({})
    lvl0 = Level0Div(t)
    for M in paths[a]:
        answer += lvl0.mobius_action(M)
    return answer.mobius_action(Hecke[a])


def Level1DivArr(t):
    """
    Returns the p + 1 level 1 divisors associated to an RM-point.
    """
    lvl0 = Level0Div(t)
    answer = []
    for a in range(p+1):
        answer.append(Level1Div(t,a))
    return answer

def Level2Div(t,h):
    """
    Return the level 2 divisor associated to an RM-point whose residues are all a(mod p). More precisely, these are the nearly reduced points equivalent to any of the double Hecke conjugates of t by a matrix of det p which also have reside a(mod p )
    """
    answer = Divisor({})
    if h!=p:
        for j in range(p):
            for M in paths[h]:
                (a,b),(c,d) = list(M)
                if (-c*j + a)%p != 0:
                    point = mobius(j,M**(-1))

                    answer += Level1Div(t,point%p).mobius_action(M)
                else: answer += Level1Div(t,p).mobius_action(M)

    elif h == p:
        for j in range(1,p+1):

            answer += Level1Div(t,j)
    return answer.mobius_action(Hecke[h])
```

```python
class MittagLeffler():

    """
    This class encodes the mittag leffler expansion of a rigid analytic function on the standard affinoid of the p-adic upper half plane

    Attributes:
        PowSeries: An array of p+1 power series in the variable v. Each power series is the expansion in the direciton of its respective edge where the local variable is 1/(z-i)
    """


    def __init__(self, PowSeries):
        self.PowSeries = PowSeries

    def __add__(self, ML):

        """
        Adds two expansions by adding their power series
        """
        addition = []
        for i in range(p+1):
            addition.append(self.PowSeries[i] + ML.PowSeries[i])
        return MittagLeffler(addition)

    def __rmul__(self, scalar):

        """
        multiplies a Mittag Leffler expansion by a scalar
        """
        answer = []
        for Series in self.PowSeries:
            answer.append(scalar*Series)
        return MittagLeffler(answer)

    def __sub__(self, ML):

        """
        Subtraction of two Mittag Leffler expansions
        """
        return self + ((-1)*ML)

    def mobius_action(self, M):

        """
        Computes the weight (-n) action of an invertible 2x2 matrix on the Mittag Leffler expansion
        """
        answer = [PS(0) for i in range(p+1)]
        (a,b), (c,d) = list(M)
        for i in range(p):
            if (a - c*i)%p != 0:
                [series, poly] = PowSerAction(self.PowSeries[i], M, i)
                answer[mobius(i,M**(-1))%p] = series
                answer[-1]+=poly

            else:
                answer[-1] += PowSerAction(self.PowSeries[i], M, i)[0]

        if c%p !=0:
            [series, poly] = PowSerAction(self.PowSeries[-1],M, "infty")
            answer[(d/-c)%p] = series
            answer[-1]+=poly
        else:
            answer[-1] += PowSerAction(self.PowSeries[-1],M, "infty")[0]
        return MittagLeffler(answer)





    def Next(self):


        """
        Computes the action of the Up operartor on the Mittag Leffler, used for generating the higher levels of the cocycles by recursion
        """
        answer = [PS(0) for i in range(p+1)]
        for i in range(p):

            translate = MittagLeffler([ PS(0) for i in range(p+1)])
            for N in paths[i]:
                translate += self.mobius_action(N**(-1))
            OldSeries = translate.PowSeries

            for center in range(p):
                P2 = p**(n)*OldSeries[center](p*v*SavedSeries[center])
                answer[i] += P2


        OldSeries = self.PowSeries
        for center in range(1,p):
            answer[-1] += remove_poly(OldSeries[center](SavedSeriesInfty[center - 1]))
        answer[-1] += remove_poly(OldSeries[-1](p*v))



        return (p**(-m))*MittagLeffler(answer)

    def poly(self):

        """
        Returns the polynomial part of the mittag leffler expansion
        """
        series_dict = self.PowSeries[-1].dict()
        poly_answer = 0
        for i in range(n+1):
            poly_answer += series_dict.get(ETuple((i,)),0)*y**i
        return poly_answer

    def differentiate(self):
        """
        Differentiates the mittag leffler expansion
        """
        answer = []
        for i in range(p):
            old_dict = self.PowSeries[i].dict()
            answer_dict = {}
            for powert, coeff in old_dict.items():
                (power, ) = powert
                answer_dict[(power + 1,)] = (-1)*power*coeff
            answer.append(PS(answer_dict))


        old_dict_infty = self.PowSeries[-1].dict()
        infty_answer_dict = {}
        for powert, coeff in old_dict_infty.items():
            (power, ) = powert
            if power!=0:
                infty_answer_dict[(power-1,)] = power*coeff
        answer.append(PS(infty_answer_dict))

        return MittagLeffler(answer)

    def iter_diff(self, i):
        """
        Iterates the differentiation
        """
        answer = self
        for j in range(i):
            answer  = answer.differentiate()
        return answer

    def fix_poly_invariance(self):
        """
        The mittag leffler expansions are often invariant up to polynomials and in certain cases this invariance can be fixed completely. This adjusts the polynomial part of the mittag leffler to be invariant completely
        """
        J1 = self.mobius_action(Sm) + self
        J2 = self.mobius_action(Um**2) + self.mobius_action(Um) + self
        c = J1.poly()[0]
        b = J2.poly()[0] - 2*c
        self.PowSeries[p] -= (b*v +c)

    def evaluate(self, t):
        """
        evaluates the expansion at a point t
        """
        total  = self.PowSeries[p](t)
        for i in range(p):
            total += self.PowSeries[i](1/(t -i))
        return(total)

    def term2(self):
        """
        Checks that the cocycle satisfies the 2 term relation up to polynomials
        """
        vanishing = (self.mobius_action(Sm) + self).min_coeff()
        print('The two term relation vanishes to a precision of ' + str(vanishing))




    def term3(self):
        """
        Checks that the cocycle satisfies the 2 term relation up to polynomials

        """
        vanishing = (self.mobius_action(Um**2) + self.mobius_action(Um) + self).min_coeff()
        print('The three term relation vanishes to a precision of ' + str(vanishing))


    def min_coeff(self):
        """
        Finds the minimal coefficient in the power-series expansion, excluding the polynomial part.
        """
        coeff_list = []
        for i in range(p):
            coeff_list += (list(self.PowSeries[i].dict().values()))

        coeff_list += list(remove_poly(self.PowSeries[p]).dict().values())
        return min([j.ordp() for j in coeff_list])
```

```python
def rCoeff(r,a,w):
    """
    Returns the contribution to the r-th coefficient of the Mittag-Leffler expansion by an RM-point w
    """
    answer = 0
    if a != p:
        for l in range(m+1):
            answer += (-1)**(r)*binomial(r,l)*binomial(n-l,m-l)*((QF1(a)-w)**(r-l))/((w-aut1(w))**(m-l))
        return -sign(w)*answer/(binomial(r,n)*(r-n)*binomial(n,m)*factorial(n))
    else:
        for l in range(m+1):
            answer += binomial(l+r-n-1,l)*binomial(n-l,m-l)/(w**(l+r-n)*(w-aut1(w))**(m-l))
        return -sign(w)*answer/((binomial(r,n)*(r-n))*binomial(n,m)*factorial(n))

def rCoeffTot(r, a, divs):
    """
    Returns the sum of contributions to the r-th coefficient of the Mittag-Leffler expansion by a set of RM points
    """
    answer = 0
    for w in divs:
        answer += rCoeff(r,a,w)
    return answer

def Level1MittagLeffler(t):
    """
    Returns the Mittag Leffler expansion of the level 1 of the recursion
    """
    ML = []
    for a in range(p+1):
        listt = (Level1Div(t,a)).DivElements()
        total = PS(0)
        for i in range(n+1,pprec):
            if a != p:
                total += hom1(rCoeffTot(i, a, listt))*v**(i-n)
            if a == p:
                total += hom1(rCoeffTot(i, a, listt))*v**i
        ML.append(total)
    return MittagLeffler(ML)

def Level2MittagLeffler(t):
    """
    Defined for test purposes. Computes by hand the level 2 expansion of the recursion
    """
    ML = []
    for a in range(p+1):
        listt = (Level2Div(t,a)).DivElements()
        total = PS(0)
        for i in range(n+1,pprec):
            if a != p:
                total += hom1(rCoeffTot(i, a, listt))*v**(i-n)
            if a == p:
                total += hom1(rCoeffTot(i, a, listt))*v**i
        ML.append(total)
    return MittagLeffler(ML)
```

```python
def remove_poly(Series):
    """
    Puts the coefficients of the polynomial part equal to 0
    """
    new_dict = Series.dict()
    for i in range(n+1):
        new_dict[ETuple((i,))] = K(0)
    return PS(new_dict)


def PowSerAction(P1,M, center):
    """
    Computes the Mobuious action of a matrix on a power series.
    """
    (a,b),(c,d) = list(M)
    det = K(M.determinant())
    detf = K(M.determinant()**(-m))
    if center == "infty":
        if c %p != 0:  ## infty to non-infty
            i = (-d/c)%p

            P2 = PS(P1(PS(a/c - v/((c**2)*(1 + (i+d/c)*v)))))



            q, r = (P2 * (c*(1+i*v)+d*v)**n).quo_rem(v**(n+1))


            poly = PS(0)

            for powert, coeff in r.dict().items():
                (power,) = powert
                if power<=n:
                    poly += coeff*(v-i)**(n-power)
            return([detf*q*v, detf*poly])
        else:

            return([detf*(c*v+d)**n*PS(P1(PS((a*v+b)/(c*v+d)))),detf*PS(0)]) ## infty to infty
    elif ((a-c*center )%p != 0): ## non-infty to non-infty
        Mcenter = mobius(center, M**(-1))
        bcenter = Mcenter%p

        P2 = PS(P1((c*(1+Mcenter*v)+d*v)*(c*Mcenter+d)/det))

        P3 = PS(P2(PS(v/(1-(Mcenter-bcenter)*v))))

        q, r = (P3*(c*(1+bcenter*v)+d*v)**n).quo_rem(v**(n+1))

        poly = PS(0)
        for powert, coeff in r.dict().items():
            (power,) = powert
            if power <=n:
                poly += coeff*(v-bcenter)**(n-power)
        return([detf* PS(q)*v,detf* poly])

    else:
        if (a-c*center)!=0:  ## non-infty to infty
            Mcenter = mobius(center, M**(-1))
            order = Mcenter.ord(p)
            sub = (c*v+d)*p**(-order)*(c*Mcenter+d)/(p**(-order)*(v-Mcenter))

            return([detf*(c*v+d)**n*P1(PS(sub)),PS(0)])
        else:
            return([detf*(c*v+d)**n*PS(P1(-c*(c*v+d))),PS(0)])


def PowSerActionOld(P1,M, center):
    """
    """
    (a,b),(c,d) = list(M)
    if center == "infty":
        if c %p != 0:  ## infty to non-infty
            i = (-d/c)%p

            P2 = PS(P1(PS(a/c - v/((c**2)*(1 + (i+d/c)*v)))))



            q, r = (P2 * (c*(1+i*v)+d*v)**n).quo_rem(v**(n+1))


            poly = PS(0)

            for powert, coeff in r.dict().items():
                (power,) = powert
                if power<=n:
                    poly += coeff*(v-i)**(n-power)
            return([q*v, poly])
        else:
            return([(c*v+d)**n*PS(P1(PS((a*v+b)/(c*v+d)))),PS(0)]) ## infty to infty
    elif ((a-c*center )%p != 0): ## non-infty to non-infty
        Mcenter = mobius(center, M**(-1))
        bcenter = Mcenter%p

        P2 = PS(P1((c*(1+Mcenter*v)+d*v)*(c*Mcenter+d)))

        P3 = PS(P2(PS(v/(1-(Mcenter-bcenter)*v))))

        q, r = (P3*(c*(1+bcenter*v)+d*v)**n).quo_rem(v**(n+1))

        poly = PS(0)
        for powert, coeff in r.dict().items():
            (power,) = powert
            if power <=n:
                poly += coeff*(v-bcenter)**(n-power)
        return([PS(q)*v,poly])

    else:
        if (a-c*center)!=0:  ## non-infty to infty
            Mcenter = mobius(center, M**(-1))
            order = Mcenter.ord(p)
            sub = (c*v+d)*p**(-order)*(c*Mcenter+d)/(p**(-order)*(v-Mcenter))

            return([(c*v+d)**n*P1(PS(sub)),PS(0)])
        else:
            return([(c*v+d)**n*PS(P1(-c*(c*v+d))),PS(0)])

```

```python
import json

def to_int(l):
    coefs = []
    for i in l:
        coefs.append([int(j) for j in i])
    return coefs


def list_to_number(l):
    """
    """
    total = 0
    for i in range(len(l)):

        if len(l[i]) == 2:
            [a,b ] = l[i]
            total += (a + b*prime_gen)*p**i
        elif len(l[i]) == 1:
            total += (a)*p**i
    return(total)

def dict_to_int(dictionary):
    """
    """
    ans = {}
    for key, mul in dictionary.items():
        ans[int(key)] = int(ZZ(mul))
    return ans

def write(cocycle, name):
    """
    Writes a Mittag Leffler expension into a json file
    """
    dictt = {}
    dictt['p'] = int(p)
    dictt['def_poly'] = dict_to_int(K.defining_polynomial().dict())

    for i in range(p+1):
        dicti = {}
        for powert, coeff in cocycle.PowSeries[i].dict().items():
            (power,) = powert
            dicti[str(power)] = [int(coeff.ordp()), to_int(list(coeff.expansion()))]
        dictt['PowSeries' + str(i)] = dicti


    x = json.dumps(dictt)
    file = open(name,'w')
    file.write(x)
    file.close()

def read(file_name):
    """
    """
    file = open(file_name,'r')
    x = json.loads(file.read())
    file.close()
    ses = []
    for i in range(p+1):
        sei_dict = {}
        for power, l in x['PowSeries'+str(i)].items():
            coeff = 0
            for j in range(len(l[1])):
                if len(l[1][j]) == 2:
                    [a,b] = l[1][j]
                    coeff += (p**j)*(a + b * prime_gen)
                elif len(l[1][j]) == 1:
                    coeff += (p**j)*(l[1][j][0])
            sei_dict[(int(power),)] = p**l[0]*coeff
        ses.append(PS(sei_dict))
    ses[0].dict()

    return(MittagLeffler(ses))
```

```python
def LMMCValue(ML, PL, s):
    """
    """
    if k == 4:
        ML_to_eval = MittagLeffler([PS(0) for i in range(p + 1)])
        PL_to_eval = PolyLog({})
        value = 0

        for M in automorph_sequence_pos(s):
            ML_to_eval += ML.mobius_action(M**(-1))
            PL_to_eval += PL.mobius_action(M**(-1))

        value += ML_to_eval.evaluate(hom3(s))/(hom3(s - aut3(s)))
        value += PL_to_eval.evaluate()/(hom3(s - aut3(s)))/4

        value += ML_to_eval.differentiate().evaluate(hom3(s))/2
        value += PL_to_eval.derivative().evaluate(hom3(s))/8
    if k == 2:

        ML_to_eval = MittagLeffler([PS(0) for i in range(p+1)])
        PL_to_eval = PolyLog({})
        j5ev4 = 0
        for M in automorph_sequence_pos(s):

            ML_to_eval += ML.mobius_action(M**(-1))
            PL_to_eval += PL.mobius_action(M**(-1))


        for i in range(p):
            j5ev4 += ML_to_eval.PowSeries[i](1/(hom1(s-i)))
        j5ev4 += ML_to_eval.PowSeries[p](hom1(s))
        j5ev4 += PL_to_eval.evaluate(hom1(s))/2
```

```python
w = var('w')
def coeffs(a):
    """
    """
    coeffs = [0,0]
    p = a.parent().prime()
    for i in range(len(a.expansion())):

        if a.expansion()[i] == []:
            pass
        else:

            coeffs[0] += (a.expansion()[i][0])*(p**(i+a.ordp()))
            if len(a.expansion()[i])==1:
                pass
            else:
                coeffs[1] += (a.expansion()[i][1])*(p**(i+a.ordp()))

    return coeffs

def algdepQp2(x,deg):
    """
    Uses the LLL-algorithm to test for algebraic relations in Q_{p^2}
    """
    M = matrix(QQ,deg+3)
    p = x.parent().prime()
    o = x.precision_absolute()
    N = p**(floor(5*o/7))
    for j in range(0,deg+1):

        M[j,j] = 1
        M[j,deg+1] = coeffs(x**j)[0]
        M[j,deg+2] = coeffs(x**j)[1]
    M[deg+1,deg+1] = N
    M[deg+2,deg+2] = N
    Y = M.LLL()

    X  = Y[0,deg+1]*Y[1]-Y[1,deg+1]*Y[0]

    P1 = X[deg+1]-X[0]
    for j in range(1,deg+1):
        P1 -= floor(X[j])*w**j

    d = gcd([P1.coefficient(w,i) for i in range(P1.degree(w)+1)])
    return P1/d

def lindepQp2(numbers):
    """
    Uses the LLL algorithm to test for linear relations in Q_{p^2}.
    """
    deg = len(numbers)
    M = matrix(QQ, deg + 2)
    p = numbers[0].parent().prime()
    o = numbers[0].parent().precision_cap()
    N = p**(floor(6.5*o/7))
    for j in range(deg):
        M[j,j] = 1
        M[j, deg] = coeffs(numbers[j])[0]
        M[j, deg+1] = coeffs(numbers[j])[1]
    M[deg,deg] = N
    M[deg + 1, deg + 1] = N
    Y = M.LLL()
    return Y
```
