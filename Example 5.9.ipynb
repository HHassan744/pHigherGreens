```python
PL5 = Level0PolyLog((1+sqrt(5))/2)
```

```python
PL20 = Level0PolyLog(sqrt(5))
```

```python
PL520 = 9 * PL5 - 2 * PL20
```

```python
PL520.fix_poly_invariance()
```

```python
s = 4*(1+rd1)/2+2

ML_to_eval = MittagLeffler([PS(0) for i in range(p+1)])
PL_to_eval = PolyLog({})

for M in automorph_sequence_pos(s):
    ML_to_eval += LMMC520.mobius_action(M**(-1))
    PL_to_eval += PL520.mobius_action(M**(-1))
```

```python
final_answer_520_80 = 0

for i in range(p):
    final_answer_520_80 += ML_to_eval.PowSeries[i](1/(hom1(s-i)))/hom1(s-aut1(s))

final_answer_520_80 += ML_to_eval.PowSeries[p](hom1(s))/hom1(s-aut1(s))
final_answer_520_80 += (PL_to_eval.evaluate(hom1(s))/(hom1(s-aut1(s))))/4

ML_to_eval_diff = ML_to_eval.differentiate()

for i in range(p):
    final_answer_520_80 += (-1) * ML_to_eval_diff.PowSeries[i](1/hom1(s-i))/2
final_answer_520_80 += (-1) * ML_to_eval_diff.PowSeries[p](hom1(s))/2
final_answer_520_80 += ((-1) * PL_to_eval.derivative().evaluate(hom1(s)))/8
```

```python
QF3.<rd3> = QuadraticField(-1)
hom3 = QF3.hom([K(-1).square_root()]) 
```

```python
rational_primes = list(primes(2,100))
quadratic_primes = []
for i in rational_primes:
    for j in QF3.primes_above(i):
        if not (QF3.ideal(i).is_prime()):
            quadratic_primes.append(j.gen(1))
log_quadratic_primes = [hom3(i).log(0) for i in quadratic_primes]
```

```python
print(lindepQp2(log_quadratic_primes + [final_answer_520_80]).str())
```
